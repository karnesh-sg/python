```python
# Cricket Over Simulator

total_runs = 0
wickets = 0
balls = []
legal_balls = 0

print("Cricket Over Simulator")
print("Enter: 0,1,2,3,4,6, W (wicket), WD (wide), NB (no-ball)\n")

while legal_balls < 6:
    ball = input(f"Ball {legal_balls + 1}: ").upper()

    if ball == "W":
        wickets += 1
        balls.append("W")
        legal_balls += 1

    elif ball == "WD":
        total_runs += 1
        balls.append("WD")

    elif ball == "NB":
        total_runs += 1
        balls.append("NB")

    elif ball.isdigit():
        runs = int(ball)
        if runs in [0, 1, 2, 3, 4, 6]:
            total_runs += runs
            balls.append(runs)
            legal_balls += 1
        else:
            print("Invalid runs! Enter 0,1,2,3,4,6.")
    else:
        print("Invalid input!")

print("\n------ SCORECARD ------")
print("Ball by Ball:", balls)
print("Runs:", total_runs)
print("Wickets:", wickets)

fours = balls.count(4)
sixes = balls.count(6)
dots = balls.count(0)

print("Fours:", fours)
print("Sixes:", sixes)
print("Dot Balls:", dots)

run_rate = total_runs / 1
print("Run Rate:", run_rate)
```

```python
import random

virat = 0
rohit = 0

for ball in range(1, 13):
    v = random.choice([0, 1, 2, 4, 6])
    r = random.choice([0, 1, 2, 4, 6])

    virat += v
    rohit += r

print("🏏 VIRAT:", virat)
print("🏏 ROHIT:", rohit)

if virat > rohit:
    print("👑 Virat wins!")
elif rohit > virat:
    print("👑 Rohit wins!")
else:
    print("🤝 It's a tie!")
```
