```python
player_names = ['Rohit', 'Kohli', 'Dhoni', 'Warner']
player_runs = [6000, 7000, 5000, 5500]

runs_dict = {name: runs for name, runs in zip(player_names, player_runs)}

print(runs_dict)
```

```python
bat={name:runs for name,runs in runs_dict.items() if runs>=5500}
bat
```
