```python
def gcd(x,y):
    a=min(x,y)
    for i in range(a,0,-1):
        if x%i==0 and y%i==0:
            print(i)
            break
```

```python
gcd(5,7)
```

```python
def si(p,t,r=12):
    return (p*r*t)/100
```

```python
si(1000,5)
```

```python
def lcm(x,y):
    a=max(x,y)
    for a in range(a,y,1):
        if x%a==0 and y%a==0:
            print(a)
            break
        
```

```python
lcm(2,4)
```

```python
def mul(a,b):
    if a==0 or b==0:
        return 0
    else:
        x=a*b
        return(x)
    
```

```python
mul(4,228)
```
