```python
# Cricket Score Calculator

total_runs = 0

print("Cricket Score Calculator")
print("------------------------")

for ball in range(1, 7):  # Loop for 6 balls (1 over)
    runs = int(input(f"Enter runs scored on ball {ball} (0,1,2,3,4,6): "))
    total_runs += runs

print("\nOver Completed!")
print("Total Runs Scored:", total_runs)
```

```python
total_runs = 0
fours = 0
sixes = 0

for ball in range(1, 7):
    runs = int(input(f"Ball {ball}: "))
    total_runs += runs

    if runs == 4:
        fours += 1
    elif runs == 6:
        sixes += 1

print("\nScorecard")
print("Total Runs:", total_runs)
print("Fours:", fours)
print("Sixes:", sixes)
```

```python
# Cricket Over Simulator

total_runs = 0
wickets = 0
balls = []
legal_balls = 0

print("Cricket Over Simulator")
print("Enter: 0,1,2,3,4,6, W (wicket), WD (wide), NB (no-ball)\n")

while legal_balls < 6:
    ball = input(f"Ball {legal_balls + 1}: ").upper()

    if ball == "W":
        wickets += 1
        balls.append("W")
        legal_balls += 1

    elif ball == "WD":
        total_runs += 1
        balls.append("WD")

    elif ball == "NB":
        total_runs += 1
        balls.append("NB")

    elif ball.isdigit():
        runs = int(ball)
        if runs in [0, 1, 2, 3, 4, 6]:
            total_runs += runs
            balls.append(runs)
            legal_balls += 1
        else:
            print("Invalid runs! Enter 0,1,2,3,4,6.")
    else:
        print("Invalid input!")

print("\n------ SCORECARD ------")
print("Ball by Ball:", balls)
print("Runs:", total_runs)
print("Wickets:", wickets)

fours = balls.count(4)
sixes = balls.count(6)
dots = balls.count(0)

print("Fours:", fours)
print("Sixes:", sixes)
print("Dot Balls:", dots)

run_rate = total_runs / 1
print("Run Rate:", run_rate)
```
