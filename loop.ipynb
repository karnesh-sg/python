```python
# Cricket Score Calculator

total_runs = 0

print("Cricket Score Calculator")
print("------------------------")

for ball in range(1, 7):  # Loop for 6 balls (1 over)
    runs = int(input(f"Enter runs scored on ball {ball} (0,1,2,3,4,6): "))
    total_runs += runs

print("\nOver Completed!")
print("Total Runs Scored:", total_runs)
```

```python
total_runs = 0
fours = 0
sixes = 0

for ball in range(1, 7):
    runs = int(input(f"Ball {ball}: "))
    total_runs += runs

    if runs == 4:
        fours += 1
    elif runs == 6:
        sixes += 1

print("\nScorecard")
print("Total Runs:", total_runs)
print("Fours:", fours)
print("Sixes:", sixes)
```
