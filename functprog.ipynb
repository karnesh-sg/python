```python
def dec(func):
    def innner():
        print("x"*50)
        func()
        print("x"*50)
    return innner
```

```python
def boo():
    print("Hello , Welcome to the world of Python")
```

```python
decor=dec(boo)
```

```python
decor()
```

```python
def arg(x,y,*args):
    r=x+y
    if args:
        r=r+sum(args)
    return r
```

```python
arg(5,6)
```

```python
arg(5,6,7)
```

```python
def create_person(name, age, gender, **extra_info):
    Person = {
        "name": name,
        "age": age,
        "gender": gender
    }
    print("kwargs has: ",extra_info)
    if extra_info:
      Person.update(extra_info)
    return Person
```

```python
create_person('Karnesh','23','Male',tools='python',role="ND")
```
