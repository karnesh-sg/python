```python
#dictionary of students

students = [
    {"name": "A", "marks": 50},
    {"name": "B", "marks": 100},
    {"name": "C", "marks": 40},
    {"name": "D", "marks": 70},
    {"name": "E", "marks": 60},
]
```

```python
val=lambda x:x['marks']
val(students[0])>val(students[1])
```

```python
sorted(students,key=lambda x:x['marks'])
```

```python
sorted(students,key=lambda x:x['marks'],reverse=True)
```
