```python
class emp:
    raise_amt=1.04
    def __init__(self,first,last,pay):
        self.first=first
        self.last=last
        self.pay=pay
        self.email=first+'.'+last+'@company.com'
    def fullname(self):
        return '{} {}'.format(self.first,self.last)
    def apply_raise(self):
        self.pay=int(self.pay*self.raise_amt)
    @classmethod
    def set_raise(cls,amt):
        cls.raise_amt=amt
    @classmethod
    def str_split(cls,emp_str):
        first,last,pay=emp_str.split('-')
        return cls(first,last,pay)
```

```python
emp.set_raise(2)
```

```python
class car:
    def __init__(self,model,year,colour,for_sale):
        self.model=model
        self.year=year
        self.colour=colour
        self.for_sale=for_sale

    def drive(self):
        print(f"you drive the {self.model}")
    def stop(self):
        print(f"you stop the {self.model}")
    def des(self):
        print(f"{self.year} {self.colour} {self.model}")
```

```python
car1=car("mustang",2024,"red",False)
print(car1.model)
print(car1.year)
print(car1.colour)
print(car1.for_sale)
```

```python
car2=car("Swift",2025,"red",True)
print(car2.model)
print(car2.year)
print(car2.colour)
print(car2.for_sale)
```

```python
car1.stop()
```

```python
car2.drive()
```

```python
car1.des()
```

```python
class animal:
    def __init__(self,name):
        self.name=name
        self.alive=True
    def eat(self):
        print(f"{self.name} is eating")
    def sleep(self):
        print(f"{self.name} is sleeping")

class dog(animal):
    pass

class cat(animal):
    pass
```

```python
dog1=dog("Zoe")
cat1=cat("garfield")

print(dog1.name)
print(cat1.alive)
dog1.eat()
dog1.sleep()

print(cat1.name)
print(cat1.alive)
cat1.eat()
cat1.sleep()
```
