```python
# Simple Cricket Score Calculator

team_name = input("Enter team name: ")
runs = int(input("Enter total runs scored: "))
wickets = int(input("Enter wickets lost: "))
overs = float(input("Enter overs played: "))

run_rate = runs / overs

print("\n------ Match Summary ------")
print("Team:", team_name)
print("Runs:", runs)
print("Wickets:", wickets)
print("Overs:", overs)
print("Run Rate:", round(run_rate, 2))

if runs >= 180:
    print("Excellent batting performance! 🏏")
elif runs >= 150:
    print("Good total on the board.")
else:
    print("Need a better batting performance.")
```
